```python
import kagglehub
import pandas as pd
import os

# Baixa o dataset
path = kagglehub.dataset_download(
    "karkavelrajaj/amazon-sales-dataset"
)

print("Arquivos encontrados:")
print(os.listdir(path))
```

```python
import pandas as pd
import os

arquivo = os.path.join(path, "amazon.csv")

df = pd.read_csv(arquivo)

print("Linhas:", df.shape[0])
print("Colunas:", df.shape[1])

df.head()
```

```python
df[df["rating"].isnull()]
```

```python
df["discounted_price"] = (
    df["discounted_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["discounted_price"] = pd.to_numeric(
    df["discounted_price"],
    errors="coerce"
)

df["actual_price"] = pd.to_numeric(
    df["actual_price"],
    errors="coerce"
)
```

```python
df["discount_percentage"] = (
    df["discount_percentage"]
    .str.replace("%", "", regex=False)
    .str.strip()
)

df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"],
    errors="coerce"
)
```

```python
df["rating"] = pd.to_numeric(
    df["rating"],
    errors="coerce"
)
```

```python
df["rating_count"] = (
    df["rating_count"]
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["rating_count"] = pd.to_numeric(
    df["rating_count"],
    errors="coerce"
)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
print("Preço com desconto:")
print(df["discounted_price"].describe())

print("\nPreço original:")
print(df["actual_price"].describe())

print("\nDesconto:")
print(df["discount_percentage"].describe())

print("\nRating:")
print(df["rating"].describe())

print("\nQuantidade de avaliações:")
print(df["rating_count"].describe())
```

```python
categorias_df = df["category"].str.split("|", expand=True)
categorias_df.columns = [f"categoria_{i+1}" for i in range(categorias_df.shape[1])]
df = pd.concat([df, categorias_df], axis=1)

df["rating_count"] = df["rating_count"].fillna(0)

df["discount_amount"] = df["actual_price"] - df["discounted_price"]

df = df.drop_duplicates(subset=["product_id"]).reset_index(drop=True)
df.to_csv("amazon_sales_cleaned.csv", index=False)
print(f"Arquivo 'amazon_sales_cleaned.csv' gerado com sucesso! Total de linhas únicas: {len(df)}")
```
